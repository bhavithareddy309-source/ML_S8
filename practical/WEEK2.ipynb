```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.backends.backend_pdf import PdfPages

sns.set_style("whitegrid")   # optional: gives cleaner-looking charts
```

```python
num_cols = ['CGPA', 'AttendancePercent', 'AptitudeTestScore',
            'CodingTestScore', 'SoftSkillsRating', 'Salary Package']
```

```python
df = pd.read_csv('placement_dataset.csv')
fig1 = plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=False, cmap="viridis", yticklabels=False)
plt.title("Missingness Heatmap - Placement Prediction Dataset")
plt.xlabel("Columns")
plt.tight_layout()
plt.show()
```

```python
fig2, axes = plt.subplots(2, 3, figsize=(15, 8))

try:
    for i, col in enumerate(num_cols):
        row, colpos = divmod(i, 3)
        sns.histplot(df[col].dropna(), bins=30, kde=True,
                     ax=axes[row, colpos], color='steelblue')
        axes[row, colpos].set_title(f'Distribution of {col}')
except KeyError as e:
    print(f"Error: {e}. The column '{e.args[0]}' was not found in the DataFrame.")
    print("Available columns in the DataFrame are:")
    print(df.columns.tolist())
    # To prevent further plotting errors with a missing column, you might want to adjust num_cols or the loop logic after inspecting the column list.

plt.tight_layout()
plt.show()
```
